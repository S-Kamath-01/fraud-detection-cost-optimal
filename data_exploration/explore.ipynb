# Data Exploration

#### This file contains the primary data exploration for the Synthetic Financial Datasets For Fraud Detection dataset

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/paysim.csv")
```

```python
print("Shape: ", df.shape)
print("\nColumns: ", list(df.columns))
```

```python
print("\n--- Class Balance ---")
print(df['isFraud'].value_counts())
print("Fraud rate: {:.4f}".format(df['isFraud'].mean()))
```

```python
print("\n---Transaction types---")
print(df['type'].value_counts())
```

```python
print("\n--- Fraud by transaction type ---")
print(df.groupby("type")["isFraud"].agg(["count", "sum", "mean"]).sort_values(by="mean", ascending=False))
```

```python
print("\n--- isFlaggedFraud vs isFraud ---")
print(pd.crosstab(df["isFraud"], df["isFlaggedFraud"]))
```

```python
print("\n--- step (time) range ---")
print("min step:", df["step"].min(), "max step:", df["step"].max())
print("(step = 1 simulated hour, so this tells us the total simulated time span)")
```

```python
print("\n--- Balance sanity check ---")
df["orig_balance_delta"] = df["oldbalanceOrg"] - df["amount"] - df["newbalanceOrig"]
print("orig_balance_delta stats:")
print(df["orig_balance_delta"].describe())
print("\nFraction of rows where orig balance delta is exactly 0:",
      (df["orig_balance_delta"] == 0).mean())
```

```python
print("\n--- Missing values ---")
print(df.isnull().sum())
```
